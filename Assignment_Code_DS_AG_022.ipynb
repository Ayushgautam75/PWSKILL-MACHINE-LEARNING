Answer: ### Question 1: What is a Convolutional Neural Network (CNN)?

A **Convolutional Neural Network (CNN)** is a type of **deep learning neural network mainly designed for processing image and visual data**. CNNs automatically learn important features such as edges, shapes, textures, and objects from images.

#### CNN Architecture

A typical CNN consists of:

1. **Convolutional Layer** – Applies filters/kernels to the image to detect features such as edges and patterns.
2. **Activation Function (ReLU)** – Introduces non-linearity into the network.
3. **Pooling Layer** – Reduces the spatial dimensions of feature maps while retaining important information.
4. **Fully Connected Layer** – Uses the extracted features to perform classification.
5. **Output Layer** – Produces the final prediction, such as *cat, dog, car,* etc.

### CNN vs Traditional Fully Connected Neural Network

| Feature                   | CNN                                                             | Fully Connected Neural Network                             |
| ------------------------- | --------------------------------------------------------------- | ---------------------------------------------------------- |
| **Architecture**          | Uses convolution, pooling, and sometimes fully connected layers | Every neuron connects to neurons in the previous layer     |
| **Image processing**      | Specifically designed for images                                | Not specifically designed for images                       |
| **Parameters**            | Fewer parameters due to local connections and weight sharing    | Very large number of parameters for high-resolution images |
| **Feature extraction**    | Automatically learns spatial features                           | Usually loses spatial relationships after flattening       |
| **Performance on images** | Generally much better                                           | Usually less efficient                                     |
| **Computation**           | More efficient for image data                                   | Computationally expensive for large images                 |

### Example

Suppose we have a **224 × 224 × 3** RGB image.

A fully connected network would first flatten it into **150,528 input values**, resulting in a huge number of connections.

A CNN instead processes small regions using filters, for example **3 × 3 kernels**, and gradually learns:

**Edges → Textures → Shapes → Objects → Final Class**

Therefore, **CNNs are preferred for image-related tasks** such as image classification, object detection, facial recognition, and medical image analysis because they preserve spatial information and require significantly fewer parameters than a traditional fully connected network.

## Question 2: Discuss the architecture of LeNet-5 and explain how it laid the foundation for modern deep learning models in computer vision.

**LeNet-5** is one of the earliest and most influential **Convolutional Neural Network (CNN)** architectures. It was developed by **Yann LeCun and his collaborators** for handwritten digit and document recognition. The architecture demonstrated that a neural network could learn useful visual features automatically from raw pixel data instead of relying entirely on manually designed features.

### Architecture of LeNet-5

The original LeNet-5 processes a **32 × 32 grayscale image** and contains a sequence of convolutional, subsampling, and fully connected layers:

**Input → C1 → S2 → C3 → S4 → C5 → F6 → Output**

| Layer      | Type                 | Main Function                             |
| ---------- | -------------------- | ----------------------------------------- |
| **Input**  | 32 × 32              | Grayscale image                           |
| **C1**     | Convolution          | Extracts low-level features such as edges |
| **S2**     | Subsampling/Pooling  | Reduces spatial dimensions                |
| **C3**     | Convolution          | Learns more complex patterns              |
| **S4**     | Subsampling/Pooling  | Further reduces feature-map size          |
| **C5**     | Convolution          | Extracts high-level features              |
| **F6**     | Fully Connected      | Combines learned features                 |
| **Output** | RBF/Classifier units | Predicts the digit/class                  |

### How LeNet-5 worked

The key idea was **hierarchical feature learning**:

**Pixels → Edges → Simple shapes → Complex shapes → Digit**

For example, when recognizing the digit **8**, early convolutional layers could detect edges and curves, while deeper layers could combine these features into the complete structure of an "8".

Unlike a traditional fully connected network, LeNet-5 used **local receptive fields and shared weights**. This dramatically reduced the number of parameters and allowed the network to exploit the spatial structure of images.

### Why LeNet-5 was important

LeNet-5 introduced several ideas that became fundamental to modern computer vision:

1. **Convolution for feature extraction**
   Instead of manually designing features, the network learned filters directly from training data.

2. **Weight sharing**
   The same convolutional filter was applied across different image locations, reducing the number of parameters.

3. **Local connectivity**
   Each neuron focused on a small region of the image rather than connecting to every pixel.

4. **Hierarchical learning**
   Multiple layers learned increasingly complex visual representations.

5. **End-to-end learning**
   The system could learn from image pixels directly to the final classification.

These principles are still present in modern CNN architectures such as **AlexNet, VGG, GoogLeNet/Inception, and ResNet**, although modern networks are considerably deeper and use improved activation functions, normalization techniques, optimization methods, and hardware acceleration.

### Contribution to modern deep learning

LeNet-5 was particularly significant because it provided an early practical demonstration that **deep, trainable convolutional architectures could outperform conventional approaches for visual recognition**. Its success on handwritten-digit/document recognition helped establish CNNs as an important approach to computer vision.

The architecture therefore became a conceptual foundation for the later CNN revolution. Modern networks essentially extend the same basic idea:

> **Convolution + hierarchical feature learning + nonlinear transformations + classification**

The major difference is that today's models can contain **dozens or hundreds of layers** and can be trained on millions of images using GPUs.

### Original Research Paper

The architecture is described in the influential paper:

**Yann LeCun, Léon Bottou, Yoshua Bengio, and Patrick Haffner, “Gradient-Based Learning Applied to Document Recognition,” *Proceedings of the IEEE*, Vol. 86, No. 11, pp. 2278–2324, 1998.** ([IEEE Xplore][1])

## Question 3: Compare and contrast AlexNet and VGGNet in terms of design principles, number of parameters, and performance. Highlight key innovations and limitations of each.

### Answer:

**AlexNet** and **VGGNet** are two landmark CNN architectures that played a major role in the development of modern computer vision. AlexNet won the **ImageNet Large Scale Visual Recognition Challenge (ILSVRC) in 2012**, while VGGNet achieved excellent results in **2014**.

| Feature                  | **AlexNet**                                | **VGGNet**                                                    |
| ------------------------ | ------------------------------------------ | ------------------------------------------------------------- |
| **Introduced**           | 2012                                       | 2014                                                          |
| **Depth**                | 8 learnable layers                         | 16–19 learnable layers                                        |
| **Convolution filters**  | Mainly 11×11, 5×5 and 3×3                  | Almost entirely 3×3                                           |
| **Parameters**           | ~60 million                                | ~138 million (VGG-16)                                         |
| **Activation**           | ReLU                                       | ReLU                                                          |
| **Pooling**              | Max pooling                                | Max pooling                                                   |
| **Major contribution**   | Demonstrated the power of deep CNNs + GPUs | Showed that deeper networks with small filters work very well |
| **ImageNet performance** | Top-5 error ≈ **15.3%**                    | Top-5 error ≈ **7.3%** for VGG-16                             |

### 1. AlexNet

**Design principles:**
AlexNet consisted of **5 convolutional layers followed by 3 fully connected layers**. It used relatively large convolutional filters in its early layers and employed overlapping max-pooling.

**Key innovations:**

* **ReLU activation**, which made training deep networks faster than using traditional sigmoid/tanh activations.
* **GPU-based training**, allowing much larger CNNs to be trained efficiently.
* **Dropout** in fully connected layers to reduce overfitting.
* **Data augmentation** to improve generalization.
* **Local Response Normalization (LRN)**, which was used in the original architecture.

**Performance:**
AlexNet achieved a dramatic improvement over previous ImageNet methods, with a **top-5 error rate of about 15.3%**, compared with approximately 26.2% for the second-place system. This result helped trigger the modern deep-learning revolution in computer vision.

**Limitations:**

* Around **60 million parameters**, making it computationally expensive.
* Large fully connected layers contributed heavily to the parameter count.
* Relatively large convolutional kernels were less efficient than the small kernels used by later architectures.
* LRN added complexity and was later largely abandoned.

---

### 2. VGGNet

**Design principles:**
VGGNet focused on a **simple and uniform architecture**. Instead of using large filters, it primarily used repeated **3×3 convolutional filters**, increasing network depth.

For example, **VGG-16** follows approximately:

**Input → Conv → Conv → Pool → Conv → Conv → Pool → Conv → Conv → Conv → Pool → FC → FC → Output**

**Key innovations:**

* Extensive use of **3×3 convolutional filters**.
* Greater **network depth**, allowing the model to learn more complex hierarchical features.
* Simple and consistent architecture that was relatively easy to understand and reproduce.
* Continued use of **ReLU and max pooling**.

**Performance:**
VGG-16 achieved a **top-5 ImageNet error of approximately 7.3%**, a substantial improvement over AlexNet. Its simple architecture also made it highly influential for transfer learning and feature extraction.

**Limitations:**

* **VGG-16 has about 138 million parameters**, more than twice AlexNet.
* Requires significant **memory and computational resources**.
* Training and inference are relatively slow compared with more efficient architectures.
* The large fully connected layers account for a large proportion of its parameters.

### Overall Comparison

The major difference is that **AlexNet demonstrated that deep CNNs could dramatically improve image recognition**, while **VGGNet showed that increasing depth with small 3×3 convolutional filters could produce even better representations**.

**AlexNet → GPU + ReLU + Dropout + Deep CNNs**

**VGGNet → Deeper architecture + Small 3×3 filters + Simple design**

Thus, AlexNet was a major breakthrough that popularized deep CNNs, whereas VGGNet refined the CNN design and established **depth and small convolutional filters** as important principles for computer vision.

## Question 4: What is Transfer Learning in the context of image classification?

### Answer:

**Transfer Learning** is a deep learning technique in which a model that has already been trained on a **large dataset** is reused for a new but related image-classification task. Instead of training a CNN completely from scratch, we use the knowledge learned by a **pre-trained model** and adapt it to our specific problem.

For example, a model such as **VGG16, ResNet, or InceptionV3** may already be trained on **ImageNet**, which contains millions of images. The early layers of these models learn general features such as **edges, colors, textures, and shapes**, while deeper layers learn more task-specific features.

### How Transfer Learning Works

A typical transfer-learning process is:

**Pre-trained Model → Remove/Replace Final Layer → Add New Classification Layer → Train on New Dataset**

For example, if VGG16 is trained to classify 1,000 ImageNet categories, we can replace its final classification layer with a layer that predicts only:

**Cat / Dog**

The pre-trained convolutional layers can initially be **frozen**, meaning their weights are not updated. Only the newly added classification layer is trained. Later, some deeper layers can be **fine-tuned** to improve performance.

### Advantages of Transfer Learning

1. **Reduces computational cost**
   Training a deep CNN from scratch requires significant computational power and time. Transfer learning starts with already learned weights, so training is much faster.

2. **Works well with limited data**
   When only a small dataset is available, training a large model from scratch can lead to **overfitting**. Pre-trained models already contain useful visual representations, allowing the model to learn effectively from fewer examples.

3. **Improves performance**
   Features learned from large datasets such as ImageNet can provide a strong starting point, often resulting in better accuracy and generalization.

4. **Requires less training time**
   Since the model does not need to learn basic visual features from the beginning, fewer training epochs may be required.

5. **Useful for specialized applications**
   Transfer learning is widely used in areas such as **medical image classification, plant disease detection, facial recognition, and object recognition**, where collecting large labeled datasets can be difficult.

### Example

Suppose we have only **2,000 images of healthy and diseased plants**.

Training a CNN from scratch:

**2,000 images → Random weights → Learn edges → Learn textures → Learn disease patterns → Classification**

With transfer learning:

**Pre-trained CNN → Already knows general visual features → Fine-tune on 2,000 plant images → Disease classification**

Therefore, transfer learning **reduces computational requirements, training time, and the amount of labeled data needed**, while often improving the accuracy and generalization of the image-classification model.

### Conclusion

Transfer learning allows us to **reuse knowledge from a pre-trained deep learning model** for a new image-classification problem. It is especially valuable when the new dataset is **small or computational resources are limited**, making it one of the most commonly used techniques in practical computer vision.

## Question 5: Describe the role of residual connections in ResNet architecture. How do they address the vanishing gradient problem in deep CNNs?

### Answer:

**Residual connections** are the key innovation of **ResNet (Residual Network)**, introduced by **Kaiming He et al. in 2015**. They allow very deep neural networks to be trained effectively by providing a shortcut path through the network.

### What is a Residual Connection?

In a traditional CNN, a few convolutional layers directly learn a mapping:

**Input → Convolution Layers → Output**

In ResNet, a **shortcut/skip connection** is added that passes the input directly to the output of the convolutional block:

**Input ───────────────→ (+) → Output**
** ↓**
**Convolution → ReLU → Convolution ─→ (+)**

Instead of learning the complete mapping **H(x)**, the network learns a **residual function F(x)**:

**H(x) = F(x) + x**

Here:

* **x** = original input
* **F(x)** = transformation learned by the convolutional layers
* **F(x) + x** = final output

### How does it solve the Vanishing Gradient Problem?

In very deep CNNs, gradients are repeatedly multiplied while being propagated backward through many layers. These gradients can become extremely small, causing the **vanishing gradient problem**. As a result, early layers learn very slowly or may stop learning effectively.

Residual connections provide a **direct path for gradient flow** from deeper layers to earlier layers.

During backpropagation, the gradient through the residual block contains a direct contribution from the shortcut connection. Therefore, even if the gradient through the convolutional layers becomes very small, the shortcut provides a stronger path for the gradient to travel backward.

### Benefits of Residual Connections

1. **Reduce vanishing gradients**
   They provide a direct path for gradients during backpropagation.

2. **Enable very deep networks**
   ResNet successfully trained networks such as **ResNet-50, ResNet-101, and ResNet-152**.

3. **Improve optimization**
   Learning a residual function is generally easier than learning a complete transformation from scratch.

4. **Improve accuracy**
   ResNet achieved excellent results on ImageNet and significantly influenced subsequent computer-vision architectures.

5. **Avoid degradation problem**
   Simply adding more layers to a conventional CNN can sometimes cause training accuracy to degrade. Residual connections help prevent this degradation.

### Simple Example

Consider a network with **100+ layers**.

Without skip connections:

**Layer 1 → Layer 2 → Layer 3 → ... → Layer 100**

The gradient has to pass through every layer.

With residual connections:

**Layer 1 ─────────────────────────→ Layer 100**
**Layer 1 → Layer 2 → ... → Layer 100**

The shortcut creates a direct route for information and gradients.

### Conclusion

Residual connections are the foundation of **ResNet's architecture**. By adding the original input directly to the output of a group of convolutional layers, they create an efficient path for information and gradients. This helps overcome the **vanishing-gradient and degradation problems**, making it possible to train much deeper CNNs with better performance.


Question 6: Implement the LeNet-5 architectures using Tensorflow or PyTorch to
classify the MNIST dataset. Report the accuracy and training time.
(Include your Python code and output in the code box below.)
Answer:

The following implementation uses TensorFlow/Keras to build a LeNet-5-style CNN and train it on the MNIST handwritten digit dataset.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import time

# --------------------------------------------------
# 1. Load MNIST dataset
# --------------------------------------------------
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize pixel values to [0, 1]
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Add channel dimension: (28, 28) -> (28, 28, 1)
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# --------------------------------------------------
# 2. Build LeNet-5 model
# --------------------------------------------------
model = models.Sequential([

    # C1 - Convolution
    layers.Conv2D(
        filters=6,
        kernel_size=(5, 5),
        activation="tanh",
        padding="same",
        input_shape=(28, 28, 1)
    ),

    # S2 - Subsampling
    layers.AveragePooling2D(pool_size=(2, 2)),

    # C3 - Convolution
    layers.Conv2D(
        filters=16,
        kernel_size=(5, 5),
        activation="tanh"
    ),

    # S4 - Subsampling
    layers.AveragePooling2D(pool_size=(2, 2)),

    # C5
    layers.Conv2D(
        filters=120,
        kernel_size=(5, 5),
        activation="tanh"
    ),

    # Flatten
    layers.Flatten(),

    # F6 - Fully Connected
    layers.Dense(84, activation="tanh"),

    # Output layer
    layers.Dense(10, activation="softmax")
])

# --------------------------------------------------
# 3. Compile model
# --------------------------------------------------
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# --------------------------------------------------
# 4. Train the model and measure training time
# --------------------------------------------------
start_time = time.time()

history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)

end_time = time.time()

training_time = end_time - start_time

# --------------------------------------------------
# 5. Evaluate the model
# --------------------------------------------------
test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nTest Accuracy:", test_accuracy * 100, "%")
print("Training Time:", round(training_time, 2), "seconds")
```

Question 7: Use a pre-trained VGG16 model (via transfer learning) on a small custom
dataset (e.g., flowers or animals). Replace the top layers and fine-tune the model.
Include your code and result discussion.
(Include your Python code and output in the code box below.)
Answer:

```python
import tensorflow as tf
import tensorflow_datasets as tfds
import time

# Ensure tensorflow-datasets is up-to-date
!pip install --upgrade tensorflow-datasets

# --------------------------------------------------
# 1. Load the Flowers dataset
# --------------------------------------------------

(train_ds, test_ds), ds_info = tfds.load(
    "tf_flowers",
    split=["train[:80%]", "train[80%:]"],
    as_supervised=True,
    with_info=True
)

num_classes = ds_info.features["label"].num_classes

print("Number of classes:", num_classes)

# --------------------------------------------------
# 2. Preprocess the images
# --------------------------------------------------

IMG_SIZE = 224
BATCH_SIZE = 32

def preprocess(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.cast(image, tf.float32)
    image = tf.keras.applications.vgg16.preprocess_input(image)

    return image, label


train_ds = (
    train_ds
    .map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)
    .shuffle(1000)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

test_ds = (
    test_ds
    .map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

# --------------------------------------------------
# 3. Load pre-trained VGG16
# --------------------------------------------------

base_model = tf.keras.applications.VGG16(
    include_top=False,
    weights="imagenet",
    input_shape=(224, 224, 3)
)

# Freeze pre-trained layers initially
base_model.trainable = False

# --------------------------------------------------
# 4. Add custom classification layers
# --------------------------------------------------

model = tf.keras.Sequential([
    base_model,

    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dense(
        256,
        activation="relu"
    ),

    tf.keras.layers.Dropout(0.5),

    tf.keras.layers.Dense(
        num_classes,
        activation="softmax"
    )
])

# --------------------------------------------------
# 5. Compile the model
# --------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# --------------------------------------------------
# 6. Train only the new top layers
# --------------------------------------------------

start_time = time.time()

history = model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=5
)

end_time = time.time()

print("\nInitial Training Time:",
      round(end_time - start_time, 2),
      "seconds")

# --------------------------------------------------
# 7. Fine-tune VGG16
# --------------------------------------------------

# Unfreeze the base model
base_model.trainable = True

# Freeze earlier layers and train deeper layers
for layer in base_model.layers[:-4]:
    layer.trainable = False

# Recompile with a very small learning rate
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.00001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Fine-tuning
history_fine = model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=5
)

# --------------------------------------------------
# 8. Evaluate the model
# --------------------------------------------------

test_loss, test_accuracy = model.evaluate(
    test_ds
)

print("\nFinal Test Accuracy:",
      round(test_accuracy * 100, 2),
      "%")
```

Question 8: Write a program to visualize the filters and feature maps of the first
convolutional layer of AlexNet on an example input image.
(Include your Python code and output in the code box below.)
Answer:

```python
import torch
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np

# --------------------------------------------------
# 1. Load pre-trained AlexNet
# --------------------------------------------------

weights = torchvision.models.AlexNet_Weights.DEFAULT
model = torchvision.models.alexnet(weights=weights)

model.eval()

# --------------------------------------------------
# 2. Get the first convolutional layer
# --------------------------------------------------

first_conv = model.features[0]

print("First Convolutional Layer:")
print(first_conv)

# Shape of filters
print("Filter shape:", first_conv.weight.shape)

# --------------------------------------------------
# 3. Visualize first-layer filters
# --------------------------------------------------

filters = first_conv.weight.detach().cpu()

fig, axes = plt.subplots(2, 4, figsize=(12, 6))

for i, ax in enumerate(axes.flat):
    filter_img = filters[i]

    # Normalize filter values to 0-1
    filter_img = filter_img - filter_img.min()
    filter_img = filter_img / filter_img.max()

    # Convert from CHW to HWC
    filter_img = filter_img.permute(1, 2, 0)

    ax.imshow(filter_img)
    ax.set_title(f"Filter {i + 1}")
    ax.axis("off")

plt.suptitle("AlexNet First Convolutional Layer Filters")
plt.show()

# --------------------------------------------------
# 4. Load example image
# --------------------------------------------------

image_path = "example.jpg"

image = Image.open(image_path).convert("RGB")

preprocess = weights.transforms()
input_tensor = preprocess(image).unsqueeze(0)

# --------------------------------------------------
# 5. Generate feature maps
# --------------------------------------------------

with torch.no_grad():
    feature_maps = first_conv(input_tensor)

print("Feature map shape:", feature_maps.shape)

# --------------------------------------------------
# 6. Visualize feature maps
# --------------------------------------------------

fig, axes = plt.subplots(2, 4, figsize=(12, 6))

for i, ax in enumerate(axes.flat):

    feature_map = feature_maps[0, i].cpu().numpy()

    ax.imshow(feature_map, cmap="viridis")
    ax.set_title(f"Feature Map {i + 1}")
    ax.axis("off")

plt.suptitle("AlexNet First Convolutional Layer Feature Maps")
plt.show()
```

Question 10: You are working in a healthcare AI startup. Your team is tasked with
developing a system that automatically classifies medical X-ray images into normal,
pneumonia, and COVID-19. Due to limited labeled data, what approach would you
suggest using among CNN architectures discussed (e.g., transfer learning with ResNet
or Inception variants)? Justify your approach and outline a deployment strategy for
production use.

Answer:

Answer:

For this healthcare AI problem, I would recommend transfer learning with a pre-trained ResNet-50 model rather than training a CNN from scratch. The dataset contains limited labeled X-ray images, so a model pre-trained on a large dataset such as ImageNet can provide useful low-level and mid-level visual features.

ResNet-50 is a strong choice because its residual connections make deep networks easier to train and reduce degradation/vanishing-gradient problems. The final classification layer can be replaced with a 3-class classifier for:

Normal
Pneumonia
COVID-19
Approach

Pre-trained ResNet-50 → Replace final layer → Freeze early layers → Train classifier → Fine-tune deeper layers → Evaluate → Deploy

Because medical datasets are often imbalanced, I would also use data augmentation, class weights, precision, recall, F1-score, sensitivity, specificity, and confusion matrix, rather than relying only on accuracy.

Python Implementation


```python
import torch
import torch.nn as nn
from torchvision import models, transforms
from torch.utils.data import DataLoader
from torchvision.datasets import ImageFolder

# --------------------------------------------------
# 1. Data preprocessing and augmentation
# --------------------------------------------------

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# --------------------------------------------------
# 2. Load dataset
# --------------------------------------------------

train_data = ImageFolder(
    "dataset/train",
    transform=train_transform
)

test_data = ImageFolder(
    "dataset/test",
    transform=test_transform
)

train_loader = DataLoader(
    train_data,
    batch_size=32,
    shuffle=True
)

test_loader = DataLoader(
    test_data,
    batch_size=32,
    shuffle=False
)

# --------------------------------------------------
# 3. Load pre-trained ResNet-50
# --------------------------------------------------

weights = models.ResNet50_Weights.DEFAULT

model = models.resnet50(weights=weights)

# Freeze pre-trained layers
for param in model.parameters():
    param.requires_grad = False

# --------------------------------------------------
# 4. Replace final classification layer
# --------------------------------------------------

num_features = model.fc.in_features

model.fc = nn.Linear(
    num_features,
    3
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = model.to(device)

# --------------------------------------------------
# 5. Loss and optimizer
# --------------------------------------------------

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model.fc.parameters(),
    lr=0.001
)

# --------------------------------------------------
# 6. Train classifier
# --------------------------------------------------

epochs = 5

for epoch in range(epochs):

    model.train()

    running_loss = 0
    correct = 0
    total = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    accuracy = 100 * correct / total

    print(
        f"Epoch {epoch+1}/{epochs}, "
        f"Loss: {running_loss:.3f}, "
        f"Accuracy: {accuracy:.2f}%"
    )

# --------------------------------------------------
# 7. Fine-tuning
# --------------------------------------------------

# Unfreeze the last ResNet block
for param in model.layer4.parameters():
    param.requires_grad = True

optimizer = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=0.00001
)

# --------------------------------------------------
# 8. Save trained model
# --------------------------------------------------

torch.save(
    model.state_dict(),
    "resnet50_xray_classifier.pth"
)

print("Model saved successfully.")
```
