6.How do we handle categorical variables in Machine Learning? What are the common techniques?
>>Machine learning models work with numbers, so categorical variables must be converted into numerical form using encoding techniques.

✅ Common Techniques to Handle Categorical Variables
1️⃣ Label Encoding

Assigns a unique number to each category

Best for ordinal data (ordered categories)

Example:

Low → 0
Medium → 1
High → 2


⚠️ Not recommended for nominal data (no order), as it may create false relationships.

2️⃣ One-Hot Encoding

Creates binary (0/1) columns for each category

Best for nominal data

Example:

Color = Red, Blue, Green
→ Red   Blue  Green
   1     0      0
   ✅ Avoids order assumption
❌ Increases dimensionality

3️⃣ Ordinal Encoding

Similar to label encoding but explicitly preserves order

Used when categories have ranking

Example:

Poor → 1
Average → 2
Good → 3

4️⃣ Target Encoding (Mean Encoding)

Replace categories with the mean of the target variable

Useful for high-cardinality features

⚠️ Risk of overfitting (use with cross-validation)

5️⃣ Binary Encoding

Converts categories to binary numbers

Reduces dimensionality compared to one-hot encoding

6️⃣ Frequency / Count Encoding

Replace categories with their frequency or count

Example:

City = Delhi (100), Mumbai (80)
🔹 Summary Table
Technique	Best Used For
Label Encoding	Ordinal data
One-Hot Encoding	Nominal data
Ordinal Encoding	Ordered categories
Target Encoding	High-cardinality data
Binary Encoding	Many categories
Frequency Encoding	Count-based importance

Accelerates SGD by adding a fraction of the previous update to the current update

Adjusts learning rate for each parameter individually

Like AdaGrad but keeps moving average of squared gradients

Solves AdaGrad’s problem of learning rate decay

Works well for RNNs and online learning

Combines Momentum + RMSProp

Maintains moving averages of gradients and squared gradients

Most commonly used optimizer in deep learning

Python Example:

optimizer = Adam(learning_rate=0.001)

23.What is sklearn.preprocessing?

Preprocessing is important because raw data may have different scales, missing values, or categorical formats.

Helps improve model performance and convergence.

Key functionalities of sklearn.preprocessing
Function / Class	Purpose
StandardScaler	Standardizes features (mean=0, std=1)

OneHotEncoder	Converts categorical variables into binary dummy variables
PolynomialFeatures	Generates polynomial features from original data
PowerTransformer	Makes data more Gaussian-like (stabilizes variance)

25.Explain data encoding?
Data encoding is the process of converting categorical (non-numeric) data into numerical values so that machine learning models can process it.

Most ML algorithms cannot work directly with text or categorical data.

Encoding helps represent categories as numbers or binary vectors.

Types of Data Encoding
Encoding Type	How it Works	Example
Label Encoding	Assigns a unique integer to each category	Red → 0, Blue → 1, Green → 2
One-Hot Encoding	Creates binary columns for each category	Red → [1,0,0], Blue → [0,1,0], Green → [0,0,1]
Ordinal Encoding	Assigns integers based on order/rank	Low → 0, Medium → 1, High → 2
Target/Mean Encoding	Encodes category using mean of target variable (used in regression/classification)	For “City” column, encode as average sale per city
Python Example
